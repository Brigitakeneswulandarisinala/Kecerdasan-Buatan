### Studi Kasus: Analisis Variabilitas Parameter Atmosfer Berdasarkan Data AWS

**R — Role**

Anda berperan sebagai ahli geofisika dan meteorologi yang berpengalaman dalam pengolahan serta interpretasi data Automatic Weather Station (AWS). Anda memahami analisis data deret waktu, karakteristik atmosfer, dan hubungan antarparameter meteorologi.

**T — Task**

Lakukan analisis terhadap data Automatic Weather Station (AWS) untuk mengetahui pola dan variabilitas kondisi atmosfer selama periode pengamatan. Analisis perubahan suhu udara, kelembapan relatif, tekanan udara, kecepatan angin, arah angin, dan curah hujan berdasarkan waktu pengamatan. Identifikasi periode ketika terjadi perubahan parameter yang signifikan serta jelaskan kemungkinan hubungan antarparameter tersebut.

**C — Context**

Data AWS merupakan data pengamatan meteorologi yang direkam secara berkala pada suatu lokasi. Data tersebut dapat digunakan untuk mengetahui karakteristik kondisi atmosfer dan perubahan cuaca dari waktu ke waktu. Analisis dilakukan untuk memperoleh gambaran mengenai pola temporal parameter atmosfer dan kondisi meteorologis yang terjadi selama periode pengamatan.

**I — Input**

Gunakan dataset AWS dalam format Excel atau CSV yang berisi informasi waktu pengamatan dan parameter meteorologi, seperti tanggal/waktu, suhu udara, kelembapan relatif, tekanan udara, kecepatan angin, arah angin, dan curah hujan apabila tersedia.

**K — Constraints**

Gunakan hanya data yang tersedia dalam dataset dan jangan membuat atau mengasumsikan nilai yang tidak terdapat dalam data. Jangan menghapus data ekstrem secara otomatis sebelum memeriksa apakah nilai tersebut merupakan kesalahan pengukuran atau kondisi atmosfer yang sebenarnya. Jika terdapat data kosong, duplikat, atau nilai yang tidak wajar, tandai dan jelaskan terlebih dahulu. Gunakan satuan sesuai dataset dan lakukan konversi hanya jika diperlukan.

**O — Output**

Output yang direncanakan berupa narasi hasil analisis data AWS yang menjelaskan karakteristik masing-masing parameter meteorologi, pola perubahan berdasarkan waktu, periode dengan nilai tertinggi dan terendah, serta hubungan antarparameter. Rencanakan pula visualisasi berupa grafik deret waktu dan diagram arah angin apabila datanya tersedia. Pada tahap ini tidak perlu membuat gambar, cukup jelaskan rancangan output yang akan dihasilkan.

**Q — Quality Criteria**

Hasil analisis harus berdasarkan data aktual, menggunakan istilah meteorologi dan geofisika secara tepat, serta menjelaskan dasar dari setiap interpretasi. Bedakan hasil pengamatan, interpretasi, dan dugaan. Jelaskan keterbatasan data apabila ditemukan.

**V — Optional Verification**

Periksa data kosong, duplikat, format tanggal dan waktu, satuan, nilai ekstrem, serta kontinuitas interval pengamatan sebelum analisis. Setelah analisis selesai, periksa kembali kesesuaian antara hasil analisis dan data yang digunakan.


### Studi Kasus: Analisis Aktivitas Gempa Berdasarkan Data Seismik

**R — Role**

Anda berperan sebagai ahli geofisika bidang seismologi yang berpengalaman dalam pengolahan katalog gempa, analisis aktivitas seismik, serta interpretasi pola spasial dan temporal kejadian gempa.

**T — Task**

Lakukan analisis terhadap data katalog gempa untuk mengetahui pola aktivitas seismik berdasarkan waktu, magnitudo, kedalaman, dan lokasi episenter. Identifikasi periode dengan aktivitas gempa yang relatif tinggi, karakteristik kedalaman gempa, serta pola persebaran episenter pada wilayah penelitian.

**C — Context**

Data gempa dapat digunakan untuk memahami karakteristik aktivitas seismik suatu wilayah. Analisis diperlukan untuk mengetahui bagaimana kejadian gempa tersebar berdasarkan waktu dan lokasi serta bagaimana magnitudo dan kedalaman gempa bervariasi selama periode pengamatan.

**I — Input**

Gunakan dataset katalog gempa yang berisi tanggal, waktu, latitude, longitude, magnitudo, dan kedalaman gempa. Jika tersedia, gunakan pula informasi lokasi, jenis gempa, atau mekanisme sumber sebagai data pendukung.

**K — Constraints**

Gunakan hanya data yang tersedia dan jangan membuat kejadian gempa baru. Jangan menentukan keberadaan sesar tertentu hanya berdasarkan persebaran episenter. Jangan menganggap gempa dengan lokasi berdekatan sebagai satu kejadian tanpa dasar yang jelas. Jika terdapat data kosong atau duplikat, identifikasi terlebih dahulu sebelum melakukan analisis.

**O — Output**

Output yang direncanakan berupa **narasi analisis aktivitas seismik** yang menjelaskan jumlah dan pola kejadian gempa berdasarkan waktu, karakteristik magnitudo dan kedalaman, serta persebaran episenter. Rencanakan visualisasi berupa grafik jumlah gempa terhadap waktu, grafik hubungan magnitudo dan kedalaman, serta peta persebaran episenter. Pada tahap ini cukup menjelaskan rancangan output tanpa membuat gambar.

**Q — Quality Criteria**

Hasil harus berdasarkan data aktual, menggunakan istilah seismologi secara tepat, dan membedakan antara pola yang terlihat pada data dengan interpretasi geofisika. Setiap kesimpulan harus memiliki dasar dari parameter yang dianalisis.

**V — Optional Verification**

Periksa data kosong, duplikasi kejadian, koordinat, magnitudo, kedalaman, dan format waktu. Periksa kembali apakah pola yang dijelaskan dalam kesimpulan benar-benar sesuai dengan dataset.



### Studi Kasus: Interpretasi Data Geolistrik Resistivitas

**R — Role**
Anda berperan sebagai ahli geofisika eksplorasi yang menguasai metode geolistrik resistivitas dan interpretasi kondisi bawah permukaan.

**T — Task**

Analisis data geolistrik untuk mengetahui variasi nilai resistivitas terhadap posisi dan kedalaman serta menginterpretasikan kemungkinan perubahan material bawah permukaan berdasarkan distribusi resistivitas.

**C — Context**

Metode geolistrik resistivitas digunakan untuk memperoleh informasi mengenai kondisi bawah permukaan berdasarkan perbedaan sifat kelistrikan material. Variasi resistivitas dapat berkaitan dengan perbedaan litologi, tingkat pelapukan, porositas, kandungan air, maupun struktur bawah permukaan.

**I — Input**

Gunakan data hasil survei geolistrik yang terdiri atas koordinat atau posisi pengukuran, nilai resistivitas, kedalaman, jarak elektroda, dan hasil inversi apabila tersedia. Gunakan pula informasi geologi daerah penelitian sebagai data pendukung jika tersedia.

**K — Constraints**

Jangan menentukan jenis batuan hanya berdasarkan satu nilai resistivitas. Pertimbangkan bahwa satu rentang resistivitas dapat dimiliki oleh beberapa jenis material. Jangan membuat interpretasi litologi apabila tidak didukung oleh data. Jika terdapat ketidakpastian, jelaskan kemungkinan interpretasi alternatif.

**O — Output**

Output yang direncanakan berupa narasi interpretasi kondisi bawah permukaan yang menjelaskan zona resistivitas rendah, sedang, dan tinggi, perubahan resistivitas terhadap kedalaman, serta kemungkinan karakteristik material pada masing-masing zona. Rencanakan visualisasi berupa penampang resistivitas dan profil perubahan resistivitas terhadap kedalaman, tetapi belum membuat gambar.

**Q — Quality Criteria**

Interpretasi harus mengikuti prinsip dasar metode geolistrik, menjelaskan hubungan antara nilai resistivitas dengan kemungkinan kondisi material, serta membedakan antara hasil pengukuran dan interpretasi. Ketidakpastian interpretasi harus dijelaskan apabila terdapat tumpang tindih nilai resistivitas.

**V — Optional Verification**

Bandingkan hasil interpretasi dengan peta geologi, kondisi geologi permukaan, data bor, informasi hidrogeologi, atau hasil survei geofisika lain apabila tersedia.


### Studi Kasus: Analisis Anomali Geomagnetik

**R — Role**
Anda berperan sebagai ahli geofisika eksplorasi yang memiliki keahlian dalam metode geomagnetik dan interpretasi anomali medan magnet bumi.

**T — Task**

Analisis data geomagnetik untuk mengidentifikasi pola anomali medan magnet secara spasial dan menginterpretasikan kemungkinan variasi sifat magnetik material bawah permukaan berdasarkan pola anomali tersebut.

**C — Context**

Metode geomagnetik memanfaatkan variasi medan magnet bumi untuk memperoleh informasi mengenai kondisi bawah permukaan. Perbedaan sifat kemagnetan batuan dapat menghasilkan anomali magnetik yang dapat membantu mengidentifikasi perubahan litologi, kontak batuan, struktur geologi, atau tubuh batuan tertentu.

**I — Input**

Gunakan data geomagnetik yang terdiri atas latitude, longitude, nilai medan magnet atau anomali magnetik, serta data elevasi jika tersedia. Gunakan peta geologi dan topografi sebagai informasi pendukung apabila tersedia.

**K — Constraints**

Jangan langsung menentukan jenis batuan hanya berdasarkan anomali magnetik tinggi atau rendah. Pertimbangkan kemungkinan pengaruh medan magnet utama, variasi harian, noise, dan proses koreksi data. Jangan menentukan struktur geologi tertentu tanpa dukungan pola data atau informasi tambahan.

**O — Output**

Output yang direncanakan berupa narasi analisis pola anomali geomagnetik yang menjelaskan zona anomali tinggi dan rendah, perubahan anomali secara spasial, serta kemungkinan sumber variasi magnetik bawah permukaan. Rencanakan visualisasi berupa peta anomali magnetik dan profil anomali pada lintasan tertentu. Gambar belum perlu dibuat pada tahap ini.

**Q — Quality Criteria**

Analisis harus menggunakan prinsip geomagnetik yang benar dan menjelaskan dasar interpretasi setiap zona anomali. Hasil harus membedakan antara pola yang berasal langsung dari data dan interpretasi mengenai kondisi bawah permukaan.

**V — Optional Verification**

Periksa koordinat, nilai medan magnet, outlier, data kosong, satuan, dan kontinuitas pengukuran. Jika tersedia, bandingkan pola anomali dengan peta geologi dan topografi untuk memeriksa konsistensi interpretasi.



### Studi Kasus: Analisis Anomali Gravitasi untuk Identifikasi Struktur Bawah Permukaan

**R — Role**
Anda berperan sebagai ahli geofisika eksplorasi yang memiliki keahlian dalam metode gravitasi dan interpretasi variasi densitas bawah permukaan.

**T — Task**

Analisis data gravitasi untuk mengidentifikasi pola anomali gravitasi dan kemungkinan variasi densitas material bawah permukaan pada wilayah penelitian. Analisis perubahan nilai anomali secara spasial dan hubungkan dengan kondisi geologi apabila data pendukung tersedia.

**C — Context**

Metode gravitasi digunakan untuk mendeteksi perubahan medan gravitasi yang berkaitan dengan perbedaan densitas material bawah permukaan. Anomali gravitasi dapat memberikan informasi mengenai perubahan litologi, cekungan, tubuh batuan berdensitas tinggi, maupun struktur geologi tertentu.

**I — Input**

Gunakan data yang terdiri atas latitude, longitude, elevasi, nilai gravitasi, serta anomali gravitasi seperti Free-Air atau Bouguer apabila tersedia. Peta geologi dan topografi dapat digunakan sebagai informasi pendukung.

**K — Constraints**

Jangan menentukan jenis batuan hanya berdasarkan nilai anomali gravitasi. Pastikan jenis anomali yang digunakan diketahui sebelum melakukan interpretasi. Jika data koreksi gravitasi belum dilakukan, jangan menganggap data tersebut sebagai anomali Bouguer. Jangan membuat interpretasi struktur yang tidak didukung oleh data.

**O — Output**

Output yang direncanakan berupa narasi interpretasi pola anomali gravitasi yang menjelaskan zona anomali tinggi dan rendah, perubahan nilai anomali secara spasial, kemungkinan variasi densitas bawah permukaan, serta hubungan dengan kondisi geologi. Rencanakan visualisasi berupa peta anomali gravitasi dan profil anomali pada lintasan tertentu, tetapi belum membuat gambar.

**Q — Quality Criteria**

Hasil analisis harus sesuai dengan prinsip metode gravitasi, menjelaskan hubungan antara anomali gravitasi dan variasi densitas secara hati-hati, serta membedakan hasil pengukuran dengan interpretasi. Keterbatasan dan ketidakpastian hasil harus disebutkan.

**V — Optional Verification**

Periksa koordinat, elevasi, satuan gravitasi, data kosong, outlier, serta jenis koreksi yang telah dilakukan. Jika tersedia, bandingkan hasil dengan peta geologi, topografi, atau data geofisika lainnya.
