```python

# ============================================================
# 1. ANALISIS AUTOMATIC WEATHER STATION (AWS)
# ============================================================


if pilihan == "1":
    print("\n=== ANALISIS VARIABILITAS PARAMETER ATMOSFER AWS ===")

    # Pastikan DataFrame tersedia
    if "df" not in globals() or df is None:
        print("ERROR: Data belum dimuat. Baca file CSV/Excel terlebih dahulu.")
    elif df.empty:
        print("ERROR: File berhasil dibaca, tetapi tidak memiliki data.")
    else:
        # Bersihkan spasi pada nama kolom
        df.columns = df.columns.astype(str).str.strip()

        print("\nKolom yang tersedia:")
        for nama_kolom in df.columns:
            print("-", repr(nama_kolom))

        # Input nama kolom
        kolom_waktu = input(
            "\nKolom tanggal/waktu: "
        ).strip()

        kolom_suhu = input(
            "Kolom suhu udara (Enter jika tidak tersedia): "
        ).strip()

        kolom_rh = input(
            "Kolom kelembapan relatif/RH (Enter jika tidak tersedia): "
        ).strip()

        kolom_tekanan = input(
            "Kolom tekanan udara (Enter jika tidak tersedia): "
        ).strip()

        kolom_angin = input(
            "Kolom kecepatan angin (Enter jika tidak tersedia): "
        ).strip()

        kolom_arah = input(
            "Kolom arah angin (Enter jika tidak tersedia): "
        ).strip()

        kolom_hujan = input(
            "Kolom curah hujan (Enter jika tidak tersedia): "
        ).strip()

        # Validasi kolom waktu
        if not kolom_waktu:
            print("ERROR: Nama kolom waktu tidak boleh kosong.")

        elif kolom_waktu not in df.columns:
            print(f"ERROR: Kolom waktu '{kolom_waktu}' tidak ditemukan.")
            print("Periksa nama kolom yang ditampilkan di atas.")

        else:
            # Salin data agar data asli tidak berubah
            data_aws = df.copy()

            # Konversi waktu
            data_aws[kolom_waktu] = pd.to_datetime(
                data_aws[kolom_waktu],
                errors="coerce"
            )

            jumlah_waktu_gagal = data_aws[kolom_waktu].isna().sum()

            print("\nJumlah waktu yang gagal dibaca:",
                  jumlah_waktu_gagal)

            # Urutkan berdasarkan waktu
            data_aws = data_aws.sort_values(
                by=kolom_waktu,
                na_position="last"
            )

            # Daftar parameter numerik
            parameter_numerik = {
                "Suhu udara": kolom_suhu,
                "Kelembapan relatif/RH": kolom_rh,
                "Tekanan udara": kolom_tekanan,
                "Kecepatan angin": kolom_angin,
                "Curah hujan": kolom_hujan
            }

            kolom_valid = {}
            kolom_tidak_ditemukan = []

            # Validasi nama kolom parameter
            for nama, kolom in parameter_numerik.items():
                if not kolom:
                    continue

                if kolom not in data_aws.columns:
                    kolom_tidak_ditemukan.append(kolom)
                    print(
                        f"PERINGATAN: Kolom {kolom!r} "
                        "tidak ditemukan dan dilewati."
                    )
                else:
                    kolom_valid[nama] = kolom

            # Konversi parameter menjadi angka
            for nama, kolom in kolom_valid.items():
                data_aws[kolom] = pd.to_numeric(
                    data_aws[kolom].astype(str).str.strip(),
                    errors="coerce"
                )

            # Analisis statistik parameter numerik
            if not kolom_valid:
                print(
                    "\nTidak ada parameter numerik yang cocok. "
                    "Periksa kembali nama kolom."
                )

            else:
                for nama, kolom in kolom_valid.items():
                    seri = data_aws[kolom].dropna()

                    print("\n" + "=" * 50)
                    print("PARAMETER:", nama)
                    print("Nama kolom:", kolom)
                    print("Jumlah data valid:", seri.count())
                    print(
                        "Jumlah data kosong/tidak numerik:",
                        data_aws[kolom].isna().sum()
                    )

                    if seri.empty:
                        print("Tidak ada data numerik yang dapat dianalisis.")
                        continue

                    print("Minimum:", seri.min())
                    print("Maksimum:", seri.max())
                    print("Rata-rata:", round(seri.mean(), 3))
                    print("Median:", round(seri.median(), 3))
                    print("Standar deviasi:", round(seri.std(), 3))

                    # Waktu minimum dan maksimum
                    idx_min = seri.idxmin()
                    idx_max = seri.idxmax()

                    print(
                        "Waktu nilai minimum:",
                        data_aws.loc[idx_min, kolom_waktu]
                    )
                    print(
                        "Waktu nilai maksimum:",
                        data_aws.loc[idx_max, kolom_waktu]
                    )

                # Analisis korelasi antarparameter
                daftar_kolom_numerik = list(
                    dict.fromkeys(kolom_valid.values())
                )

                if len(daftar_kolom_numerik) >= 2:
                    data_korelasi = data_aws[
                        daftar_kolom_numerik
                    ].dropna(how="all")

                    print("\n=== KORELASI ANTARPARAMETER ===")
                    print(
                        data_korelasi.corr().round(3)
                    )

                    print(
                        "\nCatatan: korelasi menunjukkan hubungan "
                        "statistik, bukan bukti sebab-akibat."
                    )

                # Analisis arah angin
                if kolom_arah:
                    if kolom_arah in data_aws.columns:
                        print("\n=== ANALISIS ARAH ANGIN ===")

                        arah = (
                            data_aws[kolom_arah]
                            .dropna()
                            .astype(str)
                            .str.strip()
                        )

                        if arah.empty:
                            print("Tidak ada data arah angin yang valid.")
                        else:
                            print("Jumlah data arah angin:", len(arah))
                            print("\nFrekuensi arah angin:")
                            print(arah.value_counts().head(16))

                        if kolom_angin in data_aws.columns:
                            print(
                                "\nCatatan: diagram arah angin "
                                "memerlukan pengolahan arah dan "
                                "kecepatan angin lebih lanjut."
                            )
                    else:
                        print(
                            f"\nPERINGATAN: Kolom arah angin "
                            f"{kolom_arah!r} tidak ditemukan."
                        )

                print("\n=== ANALISIS AWS SELESAI ===")
                print("Output yang dihasilkan:")
                print("- Statistik deskriptif setiap parameter.")
                print("- Waktu terjadinya nilai minimum dan maksimum.")
                print("- Matriks korelasi parameter numerik.")
                print("- Distribusi frekuensi arah angin.")
```

```python
# 3. INTERPRETASI DATA GEOLISTRIK RESISTIVITAS
# ============================================================
elif pilihan == "3":
    print("\n=== ANALISIS GEOLISTRIK RESISTIVITAS ===")
    kolom_posisi = input("Kolom posisi/jarak lintasan (Enter jika tidak tersedia): ").strip()
    kolom_kedalaman = input("Kolom kedalaman: ").strip()
    kolom_res = input("Kolom resistivitas: ").strip()

    kolom_wajib = [kolom_kedalaman, kolom_res]
    kolom_tidak_ada = [kolom for kolom in kolom_wajib if kolom not in df.columns]
    if kolom_tidak_ada:
        raise ValueError("Kolom berikut tidak ditemukan: " + ", ".join(kolom_tidak_ada))
    if kolom_posisi and kolom_posisi not in df.columns:
        print("Kolom posisi tidak ditemukan; analisis dilanjutkan tanpa kolom posisi.")
        kolom_posisi = ""

    for kolom in [kolom_kedalaman, kolom_res] + ([kolom_posisi] if kolom_posisi else []):
        df[kolom] = pd.to_numeric(df[kolom], errors="coerce")

    print("\nStatistik resistivitas:")
    print(df[kolom_res].describe())
    print("\nStatistik kedalaman:")
    print(df[kolom_kedalaman].describe())
    print("Nilai resistivitas nol atau negatif:",
          (df[kolom_res] <= 0).sum())

    # Kategori relatif berdasarkan kuartil data, bukan klasifikasi litologi universal.
    nilai_valid = df[kolom_res].dropna()
    if len(nilai_valid) > 0:
        batas_rendah = nilai_valid.quantile(1/3)
        batas_tinggi = nilai_valid.quantile(2/3)
        df["Kategori_resistivitas_rel"] = np.select(
            [df[kolom_res] <= batas_rendah, df[kolom_res] >= batas_tinggi],
            ["Relatif rendah", "Relatif tinggi"],
            default="Relatif sedang"
        )
        print("\nBatas kategori relatif berdasarkan distribusi dataset:")
        print("Rendah <= ", batas_rendah)
        print("Sedang > ", batas_rendah, " dan < ", batas_tinggi)
        print("Tinggi >= ", batas_tinggi)
        print("\nJumlah data per kategori relatif:")
        print(df["Kategori_resistivitas_rel"].value_counts())
        print("Catatan: kategori ini relatif terhadap dataset, bukan identifikasi jenis batuan.")

    if kolom_posisi:
        print("\nRancangan output: penampang posisi-kedalaman dengan warna mewakili resistivitas.")
        print("Periksa apakah data berbentuk grid/penampang sebelum membuat kontur.")
    else:
        print("\nRancangan output: profil resistivitas terhadap kedalaman.")
    print("Interpretasi litologi memerlukan data geologi/bor/hidrogeologi pendukung.")
    print("Jangan menetapkan jenis batuan hanya dari satu nilai resistivitas.")
```
